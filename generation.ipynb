```python
from docx import Document

# Read the .docx file
doc = Document("sample_text.docx")
sample_text = "\n".join([para.text for para in doc.paragraphs])
print(sample_text)
```

```python
from transformers import pipeline
import json

# Load a model (you can start with a lightweight instruction-tuned model)
generator = pipeline("text-generation", model="mistralai/Mistral-7B-Instruct-v0.2")

text = """
Machine learning is a subset of artificial intelligence that enables systems to learn from data 
and improve performance without being explicitly programmed.
"""

prompt = f"""
Generate 3 flashcards in JSON format from the following text.
Each flashcard should have a "question" and an "answer".

Text:
{sample_text}
"""

response = generator(prompt, max_new_tokens=300, temperature=0.7)
output = response[0]["generated_text"]

# Clean / parse JSON output
try:
    flashcards = json.loads(output)
except json.JSONDecodeError:
    # If the model adds extra text, you can clean it up or use regex
    print("Could not parse JSON directly.")
    print(output)
    flashcards = []

print(flashcards)
```
